```python
import whisperx
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# load model — start with a smaller model to test quickly
model = whisperx.load_model(
    "medium",           # use "medium" first to test, "large-v2" for final quality
    device,
    compute_type="float16" if device == "cuda" else "float32"
)
```

```python
# transcribe a test file - mostly pure hindi except key words
audio_path = "/content/drive/MyDrive/audios/Dhruv_Rathee__3iEDMps5P-I.mp3"

audio = whisperx.load_audio(audio_path)
result = model.transcribe(audio, batch_size=16)

# print the result
for segment in result["segments"]:
    print(segment["text"])
```

on amostly hindi channel -detected language as hindi - trascription feels good. the only inconsistency noticed is - some english words are transcribed in devnagri, while some are in roman,

#Shunyalabs model (HuggingFace)

```python
!pip install -q transformers accelerate soundfile librosa
```

```python
# Load the model
import torch
from transformers import pipeline

device = 0 if torch.cuda.is_available() else -1
torch_dtype = torch.float16 if device == 0 else torch.float32

asr = pipeline(
    "automatic-speech-recognition",
    model="shunyalabs/zero-stt-hinglish",
    device=device,
    torch_dtype=torch_dtype,
    chunk_length_s=30,      # handles audio longer than Whisper's native 30s window
    stride_length_s=5,
    generate_kwargs={"language": "hi", "task": "transcribe"},
)

print("Model loaded on:", "GPU" if device == 0 else "CPU")
```

```python
test_file = "/content/drive/MyDrive/audios/AshishChanchlanivines__cMHL43RBkJs.mp3"

result = asr(test_file)
print(result["text"])
```

COMEDY SKIT - again the same issue of weird transcription . Similar effect observed on gaming stream.

```python
test_file = "/content/drive/MyDrive/audios/Dhruv_Rathee__3iEDMps5P-I.mp3"

result = asr(test_file)
print(result["text"])
```
